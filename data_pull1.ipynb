```python
import yfinance as yf
```

```python
tickers = ["SPY", "AAPL", "MSFT", "JPM", "XOM", "JNJ"]
data = yf.download(tickers, start="2016-01-01", end="2026-11-01", auto_adjust=True)["Close"]
data.to_csv("prices.csv")
print(data.shape)
print(data.head())
```
